# Arbeiten mit strings

## Strings Zusammenfügen

Strings lassen sich auf verschiedene Weisen zusammenführen. Dieses Zusammenfügen wird im Englischen "concatenate" genannt.

Sie lassen sich zum Beispiel mit dem "+" Operator Zusammenfügen:

```
"Guten " + "Tag!"
```

Man kann die `format()` Methode nutzen:
```
tageszeit = "Tag"
"Guten {}!".format(tageszeit)
```

Oder man kann formatierte strings (f-strings) verwenden:
```
tageszeit = "Tag"
f"Guten {tageszeit}!"
```

Es ist wichtig auf die Datentypen zu achten. Zahlen, wie interger (`int`; ganze Zahl) oder floating point values (`float`; Kommazahl) lassen sich beispielsweise nicht direkt an strings anfügen.


## Strings Zerlegen

Strings lassen sich auf verschiedene Weise in ihre Elemente zerlegen. Dazu werden eckige Klammern ("\[", "\]") an den string angehängt. Nun lassen sich einzelne Elemente durch ihren Index aufrufen, indem eine ganze Zahl in den eckigen Klammern eingetragen wird:
```
text = "Toast"
print(text[3])

>>>> "s"
```

Es lassen sich auch Gruppen von Elementen ausgeben. Dazu werden zwei ganze Zahlen in die eckigen Klammern geschrieben, welche mit einem Doppelpunkt getrennt sind. Dies wird in Python "slicing" genannt.

```
text = "Toast"
print(text[1:3])

>>>> "oa"
```
Zu beachten ist dabei, dass der Elementblock (slice) vor dem letzten Element aufhört.

3. Geben Sie Blöcke mit von je drei aufeinander folgenden Elementen aus.

````python
text = "Zerlegen Sie einen string in seine Elemente."

# # Einzelne Elemente
# for element in text:
#     print(element)
    

block = 4
count = 0

while block * count <= len(text):
    
    start = count * block
    ende = (count + 1) * block
    print(text[start:ende])
    
    count += 1
    
    
````

### Aufgabe

Schreiben Sie einen Code, welcher die Summe der Zahlen von 1 bis `n` bildet.
Nutzen Sie hierfür eine while-Schleife.

Geben Sie das Ergebnis mit einem Antwortsatz aus.

Wählen Sie für `n` eine beliebige natürliche Zahl, zum Beispiel `n = 10`.


### Lösung

````python
n = 10

summe = 0
aktuelle_zahl = 1

while aktuelle_zahl <= n:
    summe = summe + aktuelle_zahl
    aktuelle_zahl += 1

print(f'Die Summe aller Zahlen bis {n} lautet: {summe}')
````

### Aufgabe

Implementieren Sie [Newtons Näherungsverfahren](https://de.wikipedia.org/wiki/Newtonverfahren) um Quadratwurzeln bis zu einer vorgegebenenen Genauigkeit iterativ zu bestimmen.

````python
# Zahl von welche die Wurzel bestimmt werden soll
zahl = 10

# Genauigkeit
toleranz = 1e-6


# Geratener Startwert (kann beispielsweise N/2 sein, oder eine andere positive Zahl)
x = zahl / 2.0


while True:
    # Newton's Verfahren
    next_x = 0.5 * (x + zahl / x)
    
    # Prüfen ob die gewünschte Toleranz erreicht ist
    if abs(x - next_x) < toleranz:
        # Toleranz wurde erreicht.
        print(f"Die Quadratwurzel von {zahl} ist etwa {next_x:.3f}")
        break
        
    # Nächste Iteration.
    x = next_x
````

### Aufgabe

Schreiben Sie ein Skript das die [Fibonacci Folge](https://de.wikipedia.org/wiki/Fibonacci-Folge) bis zur Stelle `n` berechnet. Dabei soll die Funktionsvorschrift `n` mal rekursiv ausgeführt werden. 

$$f_n = f_{n-1} + f_{n-2}$$

Startwerte sind:
```
f_1 = 1
f_2 = 2
```

Geben Sie die einzelnen Elemente der Folge mit dem print()-Befehl aus.


Wählen Sie für `n` eine beliebige natürliche Zahl, zum Beispiel `n = 10`.

### Lösung

````python
# Fibonacci Folge

wiederholungen = 10

f_1 = 1
f_2 = 1

print(f_1)
print(f_2)

for wiederholung in range(wiederholungen):
    summe = f_1 + f_2
    print(summe)
    f_1 = f_2
    f_2 = summe
    

````
